# BioInfo REST API Demo

This notebook demonstrates how to query the BioInfo REST API to aggregate genomic and structural data for a gene or protein of interest (e.g., BRCA1).

```python
import httpx
import json
from pprint import pprint
```

## Configuration
Define the API endpoint and your API key.

```python
API_BASE_URL = "http://localhost:8000"
API_KEY = "my_secure_api_key_123"

headers = {
    "X-API-Key": API_KEY
}
```

## Querying the Aggregate Endpoint
Let's query the `/api/aggregate` endpoint for `BRCA1`.

```python
search_term = "BRCA1"
response = httpx.get(f"{API_BASE_URL}/api/aggregate?term={search_term}", headers=headers)

if response.status_code == 200:
    data = response.json()
    print("Success! Data fetched:\n")
    pprint(data)
else:
    print(f"Error {response.status_code}: {response.text}")
```

## Querying Specific Endpoints
You can also query specific databases (like PDB or NCBI) directly.

```python
# Query NCBI for a specific gene
ncbi_response = httpx.get(f"{API_BASE_URL}/api/ncbi/gene/INS", headers=headers)
print("NCBI Response:")
pprint(ncbi_response.json() if ncbi_response.status_code == 200 else ncbi_response.text)
```
